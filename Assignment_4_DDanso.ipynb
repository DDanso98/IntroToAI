```python
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:
```

```python
#Create Interator for First Five Even Numbers
class EvenNumbers:
    def __init__(self, count):
        self.count = count #how many even numbers to produce
        self.current = 0 #tracks how many we've returned so far
        self.value = 0 #tracks the current even number

    def __iter__(self):
        return self #an iterator must return itself from __iter__

    def __next__(self):
        if self.current >= self.count:
            raise StopIteration #signals the loop to stop
        self.value += 2 #move to the next even number
        self.current += 1
        return self.value

#Using the iterator
even_iter = EvenNumbers(5)
for num in even_iter:
    print(num)
```

```python
#Create Generator for First Five Even Numbers
def even_number_generator(count):
    value = 0
    produced = 0
    while produced < count:
        value += 2
        yield value
        produced += 1

#Using the generator
for num in even_number_generator(5):
    print(num)
```

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
```

```python
#Use map to square all numbers in 'numbers'
squared_numbers = list(map(lambda x: x ** 2, numbers))
squared_numbers
```

```python
#Use reduce to find the product of all numbers in 'numbers'
product = reduce(lambda x, y: x * y, numbers)
product
```

```python
#Filter 'numbers' to Retrive all Even Numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
even_numbers
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
```

```python
#Create 'Rectangle' Class with Area & Permieter Methods
class Rectangle:
    def __init__(self, length, width):
        self.length = length #store the rectangle's length
        self.width = width #store the rectangle's width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

#Using the class
rect = Rectangle(5, 3)
print("Area:", rect.area())
print("Perimeter:", rect.perimeter())
```

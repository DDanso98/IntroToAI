```python
#Import Relevant Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**Task 1: Load the Titanic dataset from file provided**

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

**Task 2: Inspect the data using head(), tail(), dtypes, and describe()**

```python
#Display the first 5 rows of the dataset
df.head()
```

```python
#Display the last 5 rows of the dataset
df.tail()
```

```python
#Display the data types of each column
df.dtypes
```

```python
#Display summary statistics of the numerical columns
df.describe()
```

**Task 3: Clean and preprocess the data**

```python
#Create a new categorical feature 'PclassCategory' from 'Pclass'
df['PclassCategory'] = df['Pclass'].map({1: 'First', 2: 'Second', 3: 'Third'})
df
```

```python
#Create a new categorical feature 'CabinCategory' from 'Cabin'
df['CabinCategory'] = df['Cabin'].apply(lambda x: x[0] if pd.notnull(x) else 'Unknown')
df
```

```python
#Check for missing values in the Age column
df['Age'].isnull().sum()
```

```python
#FIll Missing Age Values (using mean age)
df.fillna(df['Age'].mean(), inplace=True)

#Confirm Successful Imputation
df['Age'].isnull().sum()
```

177 missing age values successfully filled using mean age

```python
#Create a 'Family' feature by combining 'SibSp' and 'Parch'
df['Family'] = df['SibSp'] + df['Parch']
df
```

**Task 4: Identify specific data points**

```python
#Find passengers with the highest fare
highest_fare_passengers = df[df['Fare'] == df['Fare'].max()]
highest_fare_passengers
```

The highest passenger fare is tied at $512.33

```python
#Find passengers with the largest family size
largest_family_passengers = df[df['Family'] == df['Family'].max()]
largest_family_passengers
```

The largest family size is tied at 10.

**Task 6: Multi-indexing**

```python
#Create a multi-index DataFrame grouped by 'Sex' and 'Pclass'
multi_index_df = df.groupby(['Sex', 'Pclass']).mean(numeric_only=True)
multi_index_df
```

**Task 7: Create various visualisations**

```python
#Histogram of passenger ages
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Histogram of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

As shown by the above histogram, the highest age frequency of Titanic passengers was right around 30 years old. Additionally, age density was generally highest between about 15 to 48 years old. There was also a sharp decline in passengers aged 65 and older.

```python
#Bar plot of survival rates by passenger class
survival_rates = df.groupby('Pclass')['Survived'].mean()
plt.figure(figsize=(10, 6))
sns.barplot(x=survival_rates.index, y=survival_rates.values)
plt.title('Survival Rates by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()
```

The above Bar Chart tells us that survival rates were highly dependent on passenger class. First class passengers had greater survival rates than second class and over double the rates of third class passengers. These trends likely speak to the status of individuals in first class and how they may have been prioritized during evacuation efforts. Another possibility is that first class passengers may have had more emergency resources available to them and/or easier access to lifeboats than their counterparts in second/third class.

```python
#Heatmap of correlations between numerical features
correlation_matrix = df.corr(numeric_only=True)
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

**Notable Heatmap Insights**:
*   **Pclass & Survival Rates**: -0.34 is the strongest correlation with survival among relevant features. Given that lower Pclass numbers indicate higher class, the negative symbol tells us that higher class passengers were more likely to survive - this directly supports earlier insights
*   **Fare & Survival Rates**: 0.26 / moderately positive. Passengers who paid more (often correlated with class) had somewhat better survival odds. Considering the link between fare & class, this further supports the class/status trends noted
*   **SibSP & Family**: 0.89 is the strongest relationship highlighted by the heatmap. Nonetheless, considering that the Family feature is derived from SibSP & Parch, this isn't necessarily a new insight, but rather an expected outcome

```python
#Pair plot of selected features
selected_features = ['Age', 'Fare', 'SibSp', 'Parch']
sns.pairplot(df, hue='Survived', vars=selected_features)
plt.suptitle('Pair Plot of Selected Features', y=1.02)
plt.show()
```

**Task 8: Analyze survival rates by sex and passenger class**

```python
#Calculate and display survival rates by sex
survival_by_sex = df.groupby('Sex')['Survived'].mean()
survival_by_sex
```

The survival rates for females was much higher compared to males with total rates over three times higher than males.

```python
#Calculate and display survival rates by passenger class
survival_by_class = df.groupby('Pclass')['Survived'].mean()
survival_by_class
```

As noted repeatedly in earlier findings, first class passengers had the greatest survival rates (0.63%) of all passengers, followed by second class (0.47%) and third class (0.24%).

**Task 9: Analyze age distribution using a box plot**

```python
#Create a box plot of age distribution by passenger class and sex
plt.figure(figsize=(12, 6))
sns.boxplot(x='Pclass', y='Age', hue='Sex', data=df)
plt.title('Age Distribution by Passenger Class and Sex')
plt.show()
```

As previosuly noted, the majority of passengers were roughly between the ages of 15 and 48 years old, with very few passengers being 65 years or older. Interestingly, almost all of these passengers (65 and up) were in second & third class. Other outliers include small children, also within the second & third classes.

**Task 10: Analyze fare distribution using a violin plot**

```python
#Create a violin plot of fare distribution by passenger class and survival status
plt.figure(figsize=(12, 6))
sns.violinplot(x='Pclass', y='Fare', hue='Survived', data=df)
plt.title('Fare Distribution by Passenger Class and Survival Status')
plt.show()
```

Continuing on the overall theme and relationship between class & survival, the avove violin blot reinforces this relationship as the survival rates for first class passengers were highest by far compared to second & third. Notably however, is that higher fares for first class passengers seemlingly created yet another hierarchy where survival rates are shown to increase yet again.
